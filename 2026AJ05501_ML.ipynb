```python
# ====================================================
# MACHINE LEARNING ASSIGNMENT 2
# BANK MARKETING DATASET
# ====================================================

import pandas as pd
import numpy as np
import joblib
import os
import zipfile

from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    matthews_corrcoef,
    confusion_matrix,
    classification_report
)

# ====================================================
# UPLOAD DATASET
# ====================================================

print("Upload bank.zip downloaded from UCI")

uploaded = files.upload()

# ====================================================
# EXTRACT ZIP
# ====================================================

zip_file = list(uploaded.keys())[0]

with zipfile.ZipFile(zip_file, 'r') as zip_ref:
    zip_ref.extractall()

print("ZIP extracted successfully")

# ====================================================
# FIND CSV FILE
# ====================================================

csv_file = None

for file in os.listdir():
    if file.endswith(".csv"):
        csv_file = file
        break

print("CSV Found:", csv_file)

# ====================================================
# LOAD DATASET
# ====================================================

df = pd.read_csv(csv_file, sep=';')

print("\nDataset Loaded Successfully")
print("Dataset Shape:", df.shape)

print("\nFirst 5 Rows:")
display(df.head())

print("\nDataset Information:")
df.info()

print("\nColumn Names:")
print(df.columns.tolist())
```

```python
# ====================================================
# DATA PREPROCESSING
# ====================================================

# Remove duplicates
df = df.drop_duplicates()

# Convert categorical columns to numeric
df = pd.get_dummies(df, drop_first=True)

print("Dataset Shape After Encoding:")
print(df.shape)

print("\nFirst 5 Rows:")
display(df.head())
```

```python
# ====================================================
# FEATURES AND TARGET
# ====================================================

X = df.drop("y_yes", axis=1)
y = df["y_yes"]

print("Features Shape:", X.shape)
print("Target Shape:", y.shape)
```

```python
# ====================================================
# TRAIN TEST SPLIT
# ====================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
```

```python
# ====================================================
# FEATURE SCALING
# ====================================================

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

print("Scaling Successful")
```

```python
# ====================================================
# TRAINING MODELS
# ====================================================

# Logistic Regression
logistic = LogisticRegression(max_iter=2000)
logistic.fit(X_train, y_train)

# Decision Tree
decision_tree = DecisionTreeClassifier(random_state=42)
decision_tree.fit(X_train, y_train)

# KNN
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)

# Naive Bayes
naive_bayes = GaussianNB()
naive_bayes.fit(X_train, y_train)

# SVM
svm = SVC(
    kernel='rbf',
    probability=True
)
svm.fit(X_train, y_train)

print("✅ All 5 Models Trained Successfully")
```

```python
# ====================================================
# EVALUATE MODELS
# ====================================================

def evaluate_model(model_name, model):

    pred = model.predict(X_test)

    prob = model.predict_proba(X_test)[:,1]

    return {
        "Model": model_name,
        "Accuracy": round(accuracy_score(y_test,pred),4),
        "AUC": round(roc_auc_score(y_test,prob),4),
        "Precision": round(precision_score(y_test,pred),4),
        "Recall": round(recall_score(y_test,pred),4),
        "F1": round(f1_score(y_test,pred),4),
        "MCC": round(matthews_corrcoef(y_test,pred),4)
    }

results = []

results.append(evaluate_model("Logistic Regression", logistic))
results.append(evaluate_model("Decision Tree", decision_tree))
results.append(evaluate_model("KNN", knn))
results.append(evaluate_model("Naive Bayes", naive_bayes))
results.append(evaluate_model("SVM", svm))

results_df = pd.DataFrame(results)

results_df.sort_values(
    by="Accuracy",
    ascending=False
)
```

```python
# ====================================================
# SAVE MODELS
# ====================================================

import os
import joblib

os.makedirs("models", exist_ok=True)

joblib.dump(logistic, "models/logistic.pkl")
joblib.dump(decision_tree, "models/decision_tree.pkl")
joblib.dump(knn, "models/knn.pkl")
joblib.dump(naive_bayes, "models/naive_bayes.pkl")
joblib.dump(svm, "models/svm.pkl")

joblib.dump(scaler, "models/scaler.pkl")

print("Models Saved Successfully")
```

```python
# ====================================================
# CREATE TEST DATA CSV
# ====================================================

import pandas as pd

test_df = pd.DataFrame(X_test)

test_df["y"] = y_test.reset_index(drop=True)

test_df.to_csv(
    "test_data.csv",
    index=False
)

print("test_data.csv Created Successfully")
```
